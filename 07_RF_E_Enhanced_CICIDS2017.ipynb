```python

# ============================================================
# NOTEBOOK 07 — RF-E-CIC CONSOLIDATION
# CELL 05 — FROZEN DATASET AND CLASS VERIFICATION
# ============================================================

from pathlib import Path
import pyarrow.parquet as pq
import pandas as pd

print("=" * 70)
print("CELL 05 — FROZEN CIC DATASET INTEGRITY")
print("=" * 70)

DATA_DIR = Path(
    "/content/drive/MyDrive/MSc_Project/data/CICIDS2017/processed"
)

DATASETS = {
    "Train": DATA_DIR / "cicids2017_train.parquet",
    "Validation": DATA_DIR / "cicids2017_validation.parquet",
    "Test": DATA_DIR / "cicids2017_test.parquet",
}

EXPECTED_ROWS = {
    "Train": 1764554,
    "Validation": 378119,
    "Test": 378119,
}

EXPECTED_FEATURES = 78
EXPECTED_CLASSES = 15

schemas = {}
class_distributions = {}
summary = []

# ------------------------------------------------------------
# 1. Verify Parquet metadata
# ------------------------------------------------------------

for name, path in DATASETS.items():

    print("\n" + "-" * 70)
    print(name.upper(), "PARTITION")
    print("-" * 70)

    if not path.is_file():
        print("ERROR: Dataset not found:", path)
        continue

    parquet_file = pq.ParquetFile(path)

    rows = parquet_file.metadata.num_rows
    columns = parquet_file.schema_arrow.names

    schemas[name] = columns

    # Identify target column without assuming capitalisation
    target_matches = [
        col for col in columns
        if col.strip().lower() == "label"
    ]

    if len(target_matches) != 1:
        print(
            "ERROR: Could not identify one unambiguous "
            "Label column."
        )
        print("Available columns:", columns)
        continue

    label_col = target_matches[0]

    predictors = [
        col for col in columns
        if col != label_col
    ]

    print("File:", path.name)
    print("Rows:", rows)
    print("Total columns:", len(columns))
    print("Predictors:", len(predictors))
    print("Target column:", label_col)

    row_ok = rows == EXPECTED_ROWS[name]
    feature_ok = len(predictors) == EXPECTED_FEATURES

    print(
        "Row count:",
        "PASS" if row_ok else "FAIL"
    )

    print(
        "Predictor count:",
        "PASS" if feature_ok else "FAIL"
    )

    # --------------------------------------------------------
    # 2. Read label column only for class distribution
    # --------------------------------------------------------

    label_table = pq.read_table(
        path,
        columns=[label_col]
    )

    labels = label_table.column(label_col).to_pandas()

    counts = labels.value_counts(dropna=False)
    class_distributions[name] = counts

    print("\nClass distribution:")
    print(counts.to_string())

    print("\nUnique classes:", len(counts))
    print(
        "Class count:",
        "PASS" if len(counts) == EXPECTED_CLASSES
        else "FAIL"
    )

    summary.append({
        "partition": name,
        "rows": rows,
        "features": len(predictors),
        "classes": len(counts),
        "row_check": row_ok,
        "feature_check": feature_ok,
        "class_check": len(counts) == EXPECTED_CLASSES,
    })

# ------------------------------------------------------------
# 3. Verify identical predictor schemas
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PREDICTOR SCHEMA CONSISTENCY")
print("=" * 70)

if len(schemas) == 3:

    same_schema = (
        schemas["Train"] == schemas["Validation"]
        == schemas["Test"]
    )

    print(
        "Identical column names and order:",
        "PASS" if same_schema else "FAIL"
    )

    if not same_schema:
        for name, columns in schemas.items():
            print(name, ":", columns)

else:
    print("Unable to compare all three schemas.")

# ------------------------------------------------------------
# 4. Print verification summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DATASET VERIFICATION SUMMARY")
print("=" * 70)

if summary:
    summary_df = pd.DataFrame(summary)
    print(summary_df.to_string(index=False))

all_passed = (
    len(summary) == 3
    and all(
        item["row_check"]
        and item["feature_check"]
        and item["class_check"]
        for item in summary
    )
    and len(schemas) == 3
    and schemas["Train"] == schemas["Validation"]
    == schemas["Test"]
)

print("\nOVERALL DATASET CHECK:")
print("PASS" if all_passed else "REVIEW REQUIRED")

print("\nFrozen datasets modified: NO")
print("Models retrained: NO")
print("New model created: NO")
print("=" * 70)
```

```python

# ============================================================
# NOTEBOOK 07 — RF-E-CIC CONSOLIDATION
# CELL 06 — STRICT FROZEN CONFIGURATION VERIFICATION
# READ ONLY — NO TRAINING
# ============================================================

from pathlib import Path
import pandas as pd
import math

PROJECT_ROOT = Path("/content/drive/MyDrive/MSc_Project")
SOURCE_MODEL = (
    PROJECT_ROOT / "models/RF_I1_CIC/RF_I1_S/saved_model"
)
SOURCE_RESULTS = PROJECT_ROOT / "results/RF_I1_CIC"

def single_record(filename):
    path = SOURCE_RESULTS / filename
    if not path.is_file():
        raise FileNotFoundError(path)
    df = pd.read_csv(path)
    if len(df) != 1:
        raise ValueError(
            f"{filename}: expected one record, found {len(df)}"
        )
    return df.iloc[0]

def close_to(actual, expected, tolerance=0.000005):
    return math.isclose(
        float(actual),
        float(expected),
        rel_tol=0,
        abs_tol=tolerance
    )

def is_true(value):
    return str(value).strip().lower() == "true"

def is_false(value):
    return str(value).strip().lower() == "false"

metrics = single_record("RF_I1_S_metrics.csv")
validation = single_record("RF_I1_S_validation.csv")
test = single_record("RF_I1_S_test_evaluation.csv")
selection = single_record(
    "RF_I1_imbalance_selection_decision.csv"
)

smote_df = pd.read_csv(
    SOURCE_RESULTS / "RF_I1_S_SMOTE_configuration.csv"
)
smote_config = dict(
    zip(smote_df["Parameter"], smote_df["Value"])
)

checks = []

def verify(description, condition):
    passed = bool(condition)
    checks.append((description, passed))
    print(
        f"{'PASS' if passed else 'FAIL':<6} | {description}"
    )

print("=" * 72)
print("CELL 06 — STRICT RF-E SOURCE VERIFICATION")
print("=" * 72)

print("\nMODEL AND SELECTION")
verify("SavedModel file exists",
       (SOURCE_MODEL / "saved_model.pb").is_file())
verify("SavedModel assets exist",
       (SOURCE_MODEL / "assets").is_dir())
verify("Source model is RF-I1-S",
       str(metrics["Model"]) == "RF-I1-S")
verify("Selected model is RF-I1-S",
       str(selection["Selected Imbalance Model"]) == "RF-I1-S")
verify("Reference is RF-H2-CIC",
       str(selection["Reference Model"]) == "RF-H2-CIC")
verify("Feature refinement excluded",
       is_false(metrics["Feature Refinement Included"]))
verify("Selected by validation",
       is_true(metrics["Selected By Validation"]))
verify("Selection locked before test",
       is_true(metrics["Selection Locked Before Test"]))

print("\nMODEL PARAMETERS")
verify("78 input features", int(metrics["Input Features"]) == 78)
verify("100 trees", int(metrics["Trees"]) == 100)
verify("Maximum depth = 30", int(metrics["Max Depth"]) == 30)
verify("Minimum examples = 1",
       int(metrics["Min Examples"]) == 1)
verify("Random seed = 42",
       int(metrics["Random Seed"]) == 42)

print("\nTRAINING COUNTS")
verify("Original training records = 1,764,554",
       int(metrics["Original Training Observations"]) == 1764554)
verify("SMOTE training records = 1,815,663",
       int(metrics["SMOTE Training Observations"]) == 1815663)
verify("Synthetic records = 51,109",
       int(metrics["Synthetic Observations"]) == 51109)
verify("Synthetic count is arithmetically consistent",
       int(metrics["SMOTE Training Observations"])
       - int(metrics["Original Training Observations"])
       == int(metrics["Synthetic Observations"]))

print("\nSMOTE CONFIGURATION")
verify("SMOTE method",
       str(smote_config["Method"]).upper() == "SMOTE")
verify("SMOTE k-neighbours = 3",
       int(smote_config["K Neighbours"]) == 3)
verify("StandardScaler used",
       str(smote_config["Scaling"]) == "StandardScaler")
verify("Scaler fitted on training data only",
       str(smote_config["Scaling Fit Data"])
       == "Training data only")
verify("Inverse transformation applied",
       is_true(smote_config["Inverse Transform"]))

print("\nFROZEN PERFORMANCE")
verify("Validation Macro F1 = 0.887868",
       close_to(validation["Macro F1"], 0.887868))
verify("Test Macro F1 = 0.912553",
       close_to(test["Macro F1"], 0.912553))
verify("Test accuracy = 0.998633",
       close_to(test["Accuracy"], 0.998633))
verify("Test Macro Recall = 0.927545",
       close_to(test["Macro Recall"], 0.927545))
verify("Final metrics agree with test results",
       close_to(metrics["Macro F1"], test["Macro F1"], 1e-10))
verify("Selection record agrees with validation results",
       close_to(
           selection["Selected Validation Macro F1"],
           validation["Macro F1"],
           1e-10
       ))

failed = [description for description, passed in checks
          if not passed]

print("\n" + "=" * 72)
print(f"CHECKS PASSED: {len(checks) - len(failed)}/{len(checks)}")

if failed:
    print("CELL 06: FAILED")
    raise RuntimeError(
        "Frozen configuration inconsistency: "
        + "; ".join(failed)
    )

RF_E_CONTEXT = {
    "final_model": "RF-E-CIC",
    "source_model": "RF-I1-S",
    "source_path": str(SOURCE_MODEL),
    "validation_macro_f1": float(validation["Macro F1"]),
    "test_macro_f1": float(test["Macro F1"]),
    "test_accuracy": float(test["Accuracy"]),
    "source_metrics_path": str(
        SOURCE_RESULTS / "RF_I1_S_metrics.csv"
    ),
}

print("CELL 06: PASSED")
print("Source model verified from frozen records.")
print("Model loaded or retrained: NO")
print("=" * 72)
```

```python

# ============================================================
# NOTEBOOK 07 — RF-E-CIC CONSOLIDATION
# CELL 09 — BASELINE VS FINAL ENHANCED MODEL
# ============================================================

from pathlib import Path
import pandas as pd
import math

BASELINE_FILE = (
    PROJECT_ROOT / "results/RF_B_CIC/RF_B_CIC_metrics.csv"
)

if not BASELINE_FILE.is_file():
    raise FileNotFoundError(BASELINE_FILE)

if not MANIFEST_PATH.is_file():
    raise RuntimeError("Complete Cell 08 first.")

baseline_df = pd.read_csv(BASELINE_FILE)
enhanced_df = pd.read_csv(
    SOURCE_RESULTS / "RF_I1_S_metrics.csv"
)

def normalise_key(value):
    return "".join(
        ch for ch in str(value).lower() if ch.isalnum()
    )

def select_record(df, name):
    if df.empty:
        raise ValueError(f"{name}: empty metrics file.")

    for column in df.columns:
        if normalise_key(column) == "evaluationsplit":
            test_mask = (
                df[column].astype(str)
                .str.lower()
                .str.contains("test", na=False)
            )
            if test_mask.any():
                df = df.loc[test_mask]
            break

    if len(df) != 1:
        raise ValueError(
            f"{name}: cannot identify one test metrics record. "
            f"Columns: {df.columns.tolist()}"
        )

    return df.iloc[0]

def read_metric(record, metric_name):
    alternatives = [
        metric_name,
        "Test " + metric_name,
    ]
    candidates = {normalise_key(x) for x in alternatives}

    for column in record.index:
        if normalise_key(column) in candidates:
            value = pd.to_numeric(record[column], errors="coerce")
            if not pd.isna(value):
                return float(value)

    raise KeyError(
        f"Metric '{metric_name}' not found. "
        f"Available columns: {list(record.index)}"
    )

baseline = select_record(baseline_df, "RF-B-CIC")
enhanced = select_record(enhanced_df, "RF-I1-S")

metric_names = [
    "Accuracy",
    "Macro Precision",
    "Macro Recall",
    "Macro F1",
    "Weighted F1",
    "MCC",
]

records = []

for name in metric_names:
    baseline_value = read_metric(baseline, name)
    enhanced_value = read_metric(enhanced, name)

    records.append({
        "Metric": name,
        "RF-B-CIC": baseline_value,
        "RF-E-CIC": enhanced_value,
        "Absolute Change": enhanced_value - baseline_value,
    })

comparison = pd.DataFrame(records)

baseline_f1 = read_metric(baseline, "Macro F1")
enhanced_f1 = read_metric(enhanced, "Macro F1")

if not math.isclose(
    baseline_f1, 0.893851, rel_tol=0, abs_tol=0.000005
):
    raise RuntimeError(
        "Baseline Macro F1 does not match the frozen record."
    )

if not math.isclose(
    enhanced_f1, 0.912553, rel_tol=0, abs_tol=0.000005
):
    raise RuntimeError(
        "Enhanced Macro F1 does not match the frozen record."
    )

print("=" * 72)
print("CELL 09 — RF-B-CIC VS RF-E-CIC")
print("=" * 72)

print(
    comparison.to_string(
        index=False,
        float_format=lambda x: f"{x:+.6f}"
    )
)

print("\nMACRO F1 IMPROVEMENT")
print(
    f"RF-E-CIC minus RF-B-CIC: "
    f"{enhanced_f1 - baseline_f1:+.6f}"
)

comparison_path = (
    OUTPUT_DIR / "RF_E_CIC_vs_RF_B_CIC_test_comparison.csv"
)

comparison_content = comparison.to_csv(
    index=False,
    float_format="%.9f"
).encode("utf-8")

comparison_status = write_bytes_once(
    comparison_path,
    comparison_content
)

interpretation = f"""RF-E-CIC FINAL MODEL SELECTION SUMMARY

Final identifier: RF-E-CIC
Source trained model: RF-I1-S
Dataset: CIC-IDS2017

Selection:
RF-I1-S was selected through validation Macro F1.
The final model retains 78 original predictors and controlled
SMOTE with the frozen RF-H2 tuning configuration.
Feature refinement was not selected.

Validation Macro F1: {float(validation["Macro F1"]):.9f}

Held-out CIC test:
RF-B-CIC Macro F1: {baseline_f1:.9f}
RF-E-CIC Macro F1: {enhanced_f1:.9f}
Absolute Macro F1 change: {enhanced_f1 - baseline_f1:+.9f}

Interpretive boundaries:
This difference is descriptive, not a significance-test result.
The CIC test partition was evaluated across successive frozen
experimental candidates and is not a single untouched blind
test across the entire project.
Formal paired-error testing is reserved for Notebook 08.
Very small class supports limit interpretation of rare attacks.
The final model is an alias of the saved RF-I1-S model;
no retraining or SavedModel duplication was performed.
"""

summary_path = (
    OUTPUT_DIR / "RF_E_CIC_selection_summary.txt"
)

summary_status = write_bytes_once(
    summary_path,
    interpretation.encode("utf-8")
)

print("\nOUTPUT VERIFICATION")
print("Comparison:", comparison_status)
print("Selection summary:", summary_status)
print("Comparison path:", comparison_path)
print("Summary path:", summary_path)

print("\nModel retrained: NO")
print("Frozen results overwritten: NO")
print("CELL 09: PASSED")
print("=" * 72)
```

```python

# ============================================================
# NOTEBOOK 07 — RF-E-CIC
# CELL 10 — FINAL ARTEFACT INTEGRITY VERIFICATION
# ============================================================

from pathlib import Path
import hashlib
import json
import math
import pandas as pd

PROJECT_ROOT = Path("/content/drive/MyDrive/MSc_Project")

SOURCE_MODEL = (
    PROJECT_ROOT / "models/RF_I1_CIC/RF_I1_S/saved_model"
)
SOURCE_RESULTS = PROJECT_ROOT / "results/RF_I1_CIC"
OUTPUT_DIR = PROJECT_ROOT / "results/RF_E_CIC"

ALIAS_PATH = OUTPUT_DIR / "RF_E_CIC_model_alias.json"
MANIFEST_PATH = OUTPUT_DIR / "RF_E_CIC_manifest.json"
COMPARISON_PATH = (
    OUTPUT_DIR / "RF_E_CIC_vs_RF_B_CIC_test_comparison.csv"
)
SUMMARY_PATH = OUTPUT_DIR / "RF_E_CIC_selection_summary.txt"

def file_sha256(path):
    digest = hashlib.sha256()
    with Path(path).open("rb") as handle:
        while True:
            chunk = handle.read(8 * 1024 * 1024)
            if not chunk:
                break
            digest.update(chunk)
    return digest.hexdigest()

checks = []

def verify(label, condition):
    result = bool(condition)
    checks.append((label, result))
    print(f"{'PASS' if result else 'FAIL':<6} | {label}")

print("=" * 72)
print("CELL 10 — FINAL ARTEFACT INTEGRITY VERIFICATION")
print("=" * 72)

# Verify consolidation files
print("\nCONSOLIDATION FILES")

for path in [
    ALIAS_PATH,
    MANIFEST_PATH,
    COMPARISON_PATH,
    SUMMARY_PATH,
]:
    verify(path.name, path.is_file())

if not all(result for _, result in checks):
    raise RuntimeError("Required consolidation files are missing.")

alias = json.loads(ALIAS_PATH.read_text(encoding="utf-8"))
manifest = json.loads(
    MANIFEST_PATH.read_text(encoding="utf-8")
)

# Verify model identity
print("\nMODEL IDENTITY")

verify(
    "Final identifier",
    alias["final_model_identifier"] == "RF-E-CIC"
)
verify(
    "Source identifier",
    alias["source_model_identifier"] == "RF-I1-S"
)
verify(
    "SavedModel source path",
    Path(alias["source_savedmodel_path"]) == SOURCE_MODEL
)
verify(
    "Manifest source path",
    Path(manifest["source_savedmodel_path"]) == SOURCE_MODEL
)
verify(
    "No model retraining recorded",
    manifest["retraining_performed"] is False
)
verify(
    "No SavedModel duplication recorded",
    manifest["savedmodel_copied"] is False
)

# Verify SavedModel hashes
print("\nSAVEDMODEL HASH VERIFICATION")

model_records = manifest["savedmodel_files"]
aggregate_hash = hashlib.sha256()

for relative_path, record in sorted(model_records.items()):
    path = SOURCE_MODEL / relative_path

    if not path.is_file():
        verify(f"Model file: {relative_path}", False)
        continue

    actual_hash = file_sha256(path)

    verify(
        f"Model file: {relative_path}",
        actual_hash == record["sha256"]
        and path.stat().st_size == record["size_bytes"]
    )

    aggregate_hash.update(relative_path.encode("utf-8"))
    aggregate_hash.update(bytes.fromhex(actual_hash))

verify(
    "Combined SavedModel fingerprint",
    aggregate_hash.hexdigest()
    == manifest["source_savedmodel_sha256_fingerprint"]
)

verify(
    "SavedModel file count",
    len(model_records) == manifest["savedmodel_file_count"]
)

# Verify experiment records
print("\nEXPERIMENT RECORD HASH VERIFICATION")

for filename, expected_hash in sorted(
    manifest["source_experiment_record_hashes"].items()
):
    path = SOURCE_RESULTS / filename

    verify(
        filename,
        path.is_file()
        and file_sha256(path) == expected_hash
    )

# Verify consolidated comparison
print("\nMETRIC VERIFICATION")

comparison = pd.read_csv(COMPARISON_PATH)
baseline_metrics = pd.read_csv(
    PROJECT_ROOT / "results/RF_B_CIC/RF_B_CIC_metrics.csv"
).iloc[0]
enhanced_metrics = pd.read_csv(
    SOURCE_RESULTS / "RF_I1_S_metrics.csv"
).iloc[0]

required_metrics = [
    "Accuracy",
    "Macro Precision",
    "Macro Recall",
    "Macro F1",
    "Weighted F1",
    "MCC",
]

verify(
    "Comparison contains required metrics",
    len(comparison) == len(required_metrics)
    and set(comparison["Metric"]) == set(required_metrics)
)

for metric in required_metrics:
    row = comparison.loc[
        comparison["Metric"] == metric
    ]

    if len(row) != 1:
        verify(f"{metric} comparison", False)
        continue

    row = row.iloc[0]

    baseline_value = float(baseline_metrics[metric])
    enhanced_value = float(enhanced_metrics[metric])

    consistent = (
        math.isclose(
            float(row["RF-B-CIC"]),
            baseline_value,
            abs_tol=1e-8,
            rel_tol=0
        )
        and math.isclose(
            float(row["RF-E-CIC"]),
            enhanced_value,
            abs_tol=1e-8,
            rel_tol=0
        )
        and math.isclose(
            float(row["Absolute Change"]),
            enhanced_value - baseline_value,
            abs_tol=1e-8,
            rel_tol=0
        )
    )

    verify(f"{metric} comparison", consistent)

failed = [label for label, result in checks if not result]

print("\n" + "=" * 72)
print(f"CHECKS PASSED: {len(checks) - len(failed)}/{len(checks)}")

NOTEBOOK_07_AUDIT_PASSED = len(failed) == 0

if not NOTEBOOK_07_AUDIT_PASSED:
    print("FAILED CHECKS:")
    for item in failed:
        print(" -", item)
    raise RuntimeError("Final artefact integrity check failed.")

print("CELL 10: PASSED")
print("SavedModel fingerprint:", aggregate_hash.hexdigest())
print("=" * 72)
```
